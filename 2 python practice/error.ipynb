```python
import numpy as np
import pandas as pd
```

```python
a = []

for num in range(1, 1001):
    a.append(np.random.randint(1,6))
```

```python
a = pd.Series(a)
a
```

```python
a.value_counts()
```

```python
temp = pd.Series(sorted((a.value_counts() / a.value_counts().sum()), reverse=True))
temp
```

```python
(a.value_counts() / a.value_counts().sum()).sort_index()
```

```python
temp.plot(kind='bar')
```
