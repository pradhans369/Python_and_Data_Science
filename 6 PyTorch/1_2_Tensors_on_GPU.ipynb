# **PyTorch Basics**

```python
import torch as tr
import numpy as np
```

```python
print(tr.__version__)
```

```python
if tr.cuda.is_available():
    print('GPU IS AVAILABLE')
    print(f'USING GPU : {tr.cuda.get_device_name(0)}')
else:
    print('GPU NOT AVAILABLE')
```

```python
# making a tensor on gpu
device = tr.device('cuda')

x = tr.rand(5,3, device=device)
y = tr.rand(5,3, device=device)

print(x)
print()
print(y)
```

```python
# sending a non-gpu tensor into gpu 
x = tr.rand(4,4)
print(x)
print()

temp = x.clone().to(device)                   # we use "to(device)" to add non-gpu tensor into the gpu
temp
```

#### Reshape

```python
print(temp.flatten())
print()
print(temp.reshape(2,2,2,2))
```
