#### autograd

Autograd is used to calculate gradients (used in ML and DL models).

```python
import torch as tr
import numpy as np
```

```python
# checking for GPU availability

if tr.cuda.is_available():
    print('GPU IS AVAILABLE')
    print(f'USING GPU : {tr.cuda.get_device_name(0)}')
else:
    print('GPU NOT AVAILABLE')
```

```python
x = tr.randn(5, requires_grad=True)             # adding gradient
x
```
