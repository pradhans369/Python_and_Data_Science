Tensor is a specialized multi-dimensional array designed for mathematical and computational efficiency.

- Scalars 
    - 0 dimensial tensor (a single number)
- Vectors 
    - 1 dimensioanl tensor (a list of numbers)
- Matrics
    - 2 dimensional tensors (a 2D grid of numbers)
- 3D Tensors
- 4D Tensors
    - Batches of RGB images
- 5D Tensors
    - Video data (adds a time dimension)


```python
import torch as tr
```

```python
x = tr.empty(2,2)                   # returns a tensor
y = tr.empty(3,2,2)
z = tr.empty(2,3,4,2)

print(x)
print('---------------------------------------------------------------')
print(y)
print('---------------------------------------------------------------')
print(z)
```

#### Int and float datatype tensors

```python
x = tr.ones(3,4, dtype=tr.int)
z = tr.ones(3,4, dtype=tr.float)


print(x)
print()
print(z)
```

```python
# making 1D tensor
print(tr.tensor([3,5]))

print()

# making 2D tensor
print(tr.tensor([
    [3,2],
    [5,2]
]))
```

```python
# making random tensors in int
print(tr.randint(0,11, (3,2)))              # 0, and 11 is choosing the numbers btwn 0 and 10
print()
print(tr.randint(0,10,(2,3,4)))
```

```python
# making random tensors in float
print(tr.rand(5,2))                 # does not take datatype as int
print()
print(tr.rand(2,5,3))
```

```python
x = tr.rand(2,5,3)
y = tr.rand(2,5,3)
```

```python
print(x)
print('------------------------------------------------')
print(y)
print('------------------------------------------------')
```

```python
print(x + y)
print('------------------------------------------------')
print(x + 4)
```

#### Reshaping the tensors

```python
x = tr.randint(0,10,(3,3,4))
print(x)
print('--------------------------------------------')
print()

# making a 1D tensor
print(x.view(-1))
print('--------------------------------------------')
print()

# making a 2D tensor
print(x.view(-1, 36))
print('--------------------------------------------')
print()

# 2D tensor containing 3 rows
print(x.view(3, -1))
```
