# **PyTorch Basics**

```python
import torch as tr
import numpy as np
```

```python
print(tr.__version__)
```

```python
if tr.cuda.is_available():
    print('GPU IS AVAILABLE')
    print(f'USING GPU : {tr.cuda.get_device_name(0)}')
else:
    print('GPU NOT AVAILABLE')
```

```python
print(tr.arange(0,11, 2))                             # same as np.arange
print(tr.arange(0,11))
```

```python
print(tr.linspace(1,10, 20))                        # making 20 numbers btwn 1 to 10 while every number being evenly spaced, same as np.linspace
```

```python
print(tr.eye(5))
print()
print(tr.eye(5,2))
```

```python
x = tr.randint(0,11, (3,5))
print(x.shape)
print(x.size)
```

```python
x = tr.rand(5,3)
print(x)
print()
print(tr.mean(x))                  # returns mean of all numbers in the matrics
print(tr.mean(x, dim=0))          # mean of all columns
print(tr.mean(x, dim=1))          # mean of all rows
```

```python
print(tr.median(x))
print()
print(tr.median(x, dim=0))
print()
print(tr.median(x, dim=1))
```

```python
print(tr.max(x))
print()
print(tr.max(x, dim=0))
print()
print(tr.max(x, dim=1))
```

```python
print(tr.min(x))
print()
print(tr.min(x, dim=0))
print()
print(tr.min(x, dim=1))
```

```python
print(tr.std(x))             # standard deviation
print()
print(tr.std(x, dim=0))
print()
print(tr.std(x, dim=1))
```

```python
x = tr.rand(3)
y = tr.rand(3).T
print(x)
print()
print(y)
print()
print(tr.dot(x,y))            # remember is pytorch dot product is only limited to 1D tensors
# if you want to do dot product btwn 2D or 3D tensors, then you have to use "np.dot"
```

```python
x = tr.rand(5,3)
y = tr.rand(5,3).T
print(x)
print()
print(y)
print()
print(np.dot(x,y))
```

```python
x = tr.rand(5,3)
y = tr.rand(5,3)
print(x)
print()
print(y)
print()
print(x > y)        # making comparison btwn elements
```

```python
print(x)
print()
print(tr.log(x))
print()
print(tr.exp(x))
print()
print(tr.sqrt(x))
print()
print(tr.sigmoid(x))
print()
print(tr.softmax(x, dim=0))
print()
print(tr.relu(x))
```

#### Making inplace operation

```python
x = tr.randint(0,10, (5,3))
y = tr.randint(0,10, (5,3))

print(x)
print()
print(y)
print()

x.add_(y)     # inplace operation is done to save memory usage
```

```python
# copying one tensor into another

print(x)
print()

temp = x.clone()

print(temp)
```
