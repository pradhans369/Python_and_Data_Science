```python
import tensorflow as tf
```

```python
# creating 1D tensor
x = tf.random.uniform((10,), minval=1, maxval=11, dtype=tf.int32)
x
```

```python
print(x[:])
print()
print(x[1:5])
print()
print(x[-1])
```

```python
print(x[::2])                   # printing one element, skipping next and so on
print()
print(x[::-1])                  # printing reversed
```

```python
x = tf.random.uniform((3,3,5), minval=1, maxval=10, dtype=tf.int32)
x
```

```python
print(x[0])
print()
print(x[1])
print()
print(x[0, 0])
print(x[0, 1])
print()
print(x[1,1,:4])
print(x[1,1,2])
print(x[1,1,-1])
print()
```

```python
x = tf.reshape(tf.range(12), (3,4))
x
```

```python
import numpy as np

print(tf.transpose(x))
print()
print(tf.transpose(x, perm=[1,0]))
print()
print(np.transpose(x))
```
