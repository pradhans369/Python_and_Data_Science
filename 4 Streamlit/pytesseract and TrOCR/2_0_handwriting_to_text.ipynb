### Handwritten text image to text using `TrOCR`

- Microsoft trained TrOCR on individual cropped strips of text (one line at a time).
- When you feed a full photograph containing 2 or more vertical lines plus the notebook header ("MONDAY", calendar numbers), the Vision Transformer squashes all the lines together into a single 384 × 384 square and tries to predict everything as one single sentence!

```python
from PIL import Image

from transformers import RobertaTokenizer, ViTImageProcessor, TrOCRProcessor, VisionEncoderDecoderModel
```

```python
# 1. Load Tokenizer, Image Processor & Model (Bypasses the Transformers 5.x bug!)
tokenizer = RobertaTokenizer.from_pretrained("microsoft/trocr-base-handwritten")
image_processor = ViTImageProcessor.from_pretrained("microsoft/trocr-base-handwritten")
processor = TrOCRProcessor(image_processor=image_processor, tokenizer=tokenizer)

model = VisionEncoderDecoderModel.from_pretrained("microsoft/trocr-base-handwritten")

# 2. Open your handwritten image
img_path = "dataset/img_1.jpg"
image = Image.open(img_path).convert("RGB")

# 3. Convert image to tensors
pixel_values = processor(image, return_tensors="pt").pixel_values

# 4. Run handwriting prediction!
generated_ids = model.generate(pixel_values)
text = processor.batch_decode(generated_ids, skip_special_tokens=True)[0]
```
