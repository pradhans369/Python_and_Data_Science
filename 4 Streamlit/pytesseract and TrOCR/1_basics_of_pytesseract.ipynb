```python
import numpy as np
import pandas as pd

import pytesseract
from pdf2image import convert_from_path
import pypdfium2 as pdfium
```

### PDF to text using Pytesseract

```python
# opens the pdf and loads it into the memory
pdf = pdfium.PdfDocument(r"dataset\Late Delivery Risk Prediction APL Logistics.pdf")
```

```python
# This line tells Python the exact location on your hard drive where the actual tesseract.exe software is installed so it can run OCR.
# without manually installing the pytesseract on windows it will not work
pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'

text = []

for i, page in enumerate(pdf):
    img = page.render(scale=2).to_pil()

    # running OCR with pytesseract
    text.append(pytesseract.image_to_string(img))
```

- `.render(scale=2)`: Renders the PDF page into a visual pixel bitmap.
    - `scale=2`: Increases resolution by 2x (144 DPI), making tiny fonts razor-sharp so the OCR model doesn't misread letters!
- `.to_pil()`: Converts the raw rendered bitmap into a standard Python Pillow (PIL.Image) object.

```python
print(len(text))
```

```python
print(text[3])
```

### PDF to text using `pdfium`

Working with `pdfium` only works with pdfs, it is way faster compared to pytesseract for pdfs.

- In the previous method, Python took a picture of each page and used AI (Tesseract) to read the letters (8.1 seconds).
- In this method, Python directly extracts the digital text characters already typed inside the PDF file (0.0 seconds / Instantaneous!).

```python
import pypdfium2 as pdfium
```

```python
pdf = pdfium.PdfDocument(r"dataset\Late Delivery Risk Prediction APL Logistics.pdf")
```

```python
text = []

for i, page in enumerate(pdf):
    textpage = page.get_textpage()                      # It grabs the exact Unicode characters and letters directly from the PDF file format.
    text.append(textpage.get_text_range())              # Pulls all the characters on that page from the first letter to the last letter as a clean Python string.
```

```python
print(len(text))
```

```python
print(text[2])
```
