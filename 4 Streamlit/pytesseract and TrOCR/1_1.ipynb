### From Url image to text

```python
import requests
import io

url = "https://www.pdfunit.com/en/documentation/java/images/ocr-1.png"
headers = {
    "User-Agent": "Mozilla/5.0"
}
response = requests.get(url, headers=headers)
response.raise_for_status()

print(pyt.image_to_string(Image.open(io.BytesIO(response.content))))
```
